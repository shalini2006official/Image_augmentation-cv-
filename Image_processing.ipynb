```python
rotation = cv2.rotate(img,cv2.ROTATE_180)
from google.colab.patches import cv2_imshow
cv2_imshow(rotation)
```

```python
flip = cv2.flip(img,1)
from google.colab.patches import cv2_imshow
cv2_imshow(flip)

#1-vertical
#0-horizontal
```

```python
#image_bluring
blur = cv2.blur(img,(1,70))#limit 1 to 100
from google.colab.patches import cv2_imshow
cv2_imshow(blur)
```
